**EJERCICIO 11:**
Dada la lista notas = [8.5, 6.0, 9.75, 4.5, 7.0, 10.0, 5.25], calcule con un ciclo (sin usar sum, max ni min): el promedio con 2 decimales, la nota más alta, la más baja y cuántos estudiantes aprobaron (nota ≥ 7).
Escriba su código Python completo en el cuadro de respuesta. Respete la sangría de 4 espacios.

```python
notas = [8.5, 6.0, 9.75, 4.5, 7.0, 10.0, 5.25]
cantidad = len(notas)
ordenar = sorted(notas)
nota_alta = ordenar[-1]
nota_baja = ordenar[0]
suma = 0
aprobados = 0
print(f"Nota mas alta: {nota_alta}")
print(f"Nota mas baja: {nota_baja}")
for i in notas:
    suma = suma + i
    if i >= 7:
       aprobados = aprobados +1     
promedio = suma / cantidad
print(f"El promedio es: {promedio:.2f}")
print(f"Cantidad de aprobados es: {aprobados}")
```
